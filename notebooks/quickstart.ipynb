# Surface Code Benchmark Framework: Quick Start

This notebook provides a short introduction to the **Surface Code Benchmark Framework**, a modular Python framework for benchmarking quantum error correction (QEC) using surface codes.

The framework provides interchangeable simulation backends, decoding algorithms, experiment pipelines, analysis utilities, and visualization tools for evaluating the logical performance of surface codes under configurable noise models.

Rather than exploring the low-level details of Stim or Qiskit, this notebook introduces the framework's **public API** and demonstrates the typical benchmarking workflow.

## 1. Creating a Simulation Backend

The framework exposes a common backend interface for running logical-memory simulations and benchmarking experiments. Each backend implements the same public API while providing a different underlying implementation.

The available backends are:

| Backend | Purpose |
|---------|---------|
| **Stim** | Production backend based on Stim's canonical surface-code circuit generators. |
| **Checkerboard** | Explicit first-principles reference implementation for validation and educational use. |
| **Qiskit** | Lightweight reference implementation using Qiskit and Aer. |

In this tutorial, we will use the **Stim backend** with the **rotated surface code**, which is the recommended backend for benchmarking experiments.

````python
from qec.backends.stim import StimBackend

backend = StimBackend(
    lattice="rotated",
)

backend
````

## 2. Running a Logical Memory Experiment

The primary interface exposed by each backend is the `logical_failure_rate()` method. Given a code distance, noise model, and number of Monte Carlo shots, it estimates the probability that a logical error occurs during a logical-memory experiment.

In this example, we estimate the logical **Z-memory** failure rate of a distance-3 rotated surface code under identical depolarising and readout error probabilities.

````python
logical_rate = backend.logical_failure_rate(
    distance=3,
    rounds=3,
    shots=10_000,
    depolarizing_error=0.005,
    readout_error=0.005,
    memory_basis="Z",
)

print(f"Estimated logical failure rate: {logical_rate:.6e}")
````

## 3. Running Benchmark Experiments

While `logical_failure_rate()` is useful for individual simulations, the framework is primarily designed for automated benchmarking experiments.

Experiment scripts provide reproducible workflows for evaluating surface-code performance across multiple code distances, physical error rates, and backend implementations. Numerical results are automatically cached as CSV files, allowing figures to be regenerated without repeating expensive Monte Carlo simulations.

The current benchmark suite includes:

- Logical failure-rate scaling
- Threshold estimation
- Runtime benchmarking
- Rotated vs. unrotated lattice comparison
- Stim vs. checkerboard backend comparison
- Qiskit decoder strategy comparison

## 4. Inspecting the Results

Each benchmark experiment automatically generates figures and caches the corresponding numerical results as CSV files. This enables visualizations to be regenerated and analyses to be repeated without rerunning the underlying Monte Carlo simulations.

Generated outputs are organised by backend and experiment under the `results/` directory. For example:

```text
results/
└── stim/
    ├── threshold/
    ├── distance_scaling/
    ├── runtime_benchmark/
    └── compare_rotated_unrotated/
```

Each experiment directory typically contains:

- benchmark figures (`.png`),
- cached numerical results (`.csv`).

This output structure supports reproducible benchmarking while keeping experiment scripts independent of data analysis and visualization.
